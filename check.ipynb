```python
from fakecephus_108q.qiskit import FakeCephus108Q
from fakecephus_108q.specs import CEPHUS_108Q_SPECS_CALIBRATED
from qiskit import QuantumCircuit

# Create backend with realistic per-qubit calibration noise
backend = FakeCephus108Q(specs=CEPHUS_108Q_SPECS_CALIBRATED, noise=True)
ideal_backend = FakeCephus108Q(specs=CEPHUS_108Q_SPECS_CALIBRATED, noise=False)

# Create and run a circuit
qc = QuantumCircuit(10)
qc.h(0)
for i in range(9):
    qc.cx(i, i+1)
qc.measure_all()

# Run the same circuit on the ideal backend
ideal_job = ideal_backend.run(qc, shots=1024)
ideal_result = ideal_job.result()
print("Ideal backend counts:", ideal_result.get_counts())

job = backend.run(qc, shots=1024)
result = job.result()
print("Noisy backend counts:", result.get_counts())
```
